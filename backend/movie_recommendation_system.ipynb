```python
import numpy as np
import pandas as pd
```

```python
movies = pd.read_csv('tmdb_5000_movies1.csv')
credits = pd.read_csv('tmdb_5000_credits1.csv')
```

```python
movies = movies.merge(credits,on='title')
```

```python
movies.head(1)
```

```python
movies = movies[['movie_id','title','overview','genres','keywords','cast','crew']]
                 
```

```python
movies.isnull().sum()
```

```python
movies.dropna(inplace=True)
```

```python
movies.isnull().sum()
```

```python
movies.duplicated().sum()
```

```python
movies['genres'].head()
```

```python
import ast

def convert(obj):
    L = []

    for i in ast.literal_eval(obj):
        L.append(i['name'])

    return L

movies['genres'] = movies['genres'].apply(convert)
```

```python
movies['genres']
```

```python
movies['keywords'] = movies['keywords'].apply(convert)
```

```python
movies['keywords'].head()
```

```python
def convert3(obj):
    L = []

    counter = 0
    for i in ast.literal_eval(obj):
        if counter != 3:
            L.append(i['name'])
            counter += 1
        else:
            break
    return L
```

```python
movies['cast'] = movies['cast'].apply(convert3)
```

```python
movies['cast'].head()
```

```python
def fetch_dir(obj):
    L = []
    for i in ast.literal_eval(obj):
        if i['job'] == 'Director':
            L.append(i['name'])
    return L
```

```python
movies['crew'] = movies['crew'].apply(fetch_dir)
```

```python
movies['crew']
```

```python
movies['overview'] = movies['overview'].apply(lambda x:x.split())
```

```python
movies['overview']
```

```python
movies['genres'] = movies['genres'].apply(lambda x: [i.replace(" ","") for i in x])
```

```python
movies['genres']
```

```python
movies['keywords'] = movies['keywords'].apply(lambda x: [i.replace(" ","") for i in x])
```

```python
movies['cast'] = movies['cast'].apply(lambda x: [i.replace(" ","") for i in x])
```

```python
movies['crew'] = movies['crew'].apply(lambda x: [i.replace(" ","") for i in x])
```

```python
movies.head()
```

```python
movies['tags'] =  movies['genres'] + movies['keywords'] + movies['overview'] + movies['cast'] + movies['crew']
```

```python
cleaned_movies = movies[['movie_id','title','tags']].copy()
```

```python
cleaned_movies['tags'] = cleaned_movies['tags'].apply(lambda x: ' '.join(x))
```

```python
cleaned_movies['tags'][0]
```

```python
cleaned_movies['tags'] = cleaned_movies['tags'].apply(lambda x:x.lower())
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features=5000,stop_words='english')
```

```python
vectors = cv.fit_transform(cleaned_movies['tags']).toarray()
```

```python
vectors[999]
```

```python
cv.get_feature_names_out()
```

```python
import nltk
```

```python
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
```

```python
def stem(text):
    y = []
    for i in text.split():
        y.append(ps.stem(i))
    return " ".join(y)
```

```python
cleaned_movies['tags'] = cleaned_movies['tags'].apply(stem)
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity = cosine_similarity(vectors)
```

```python
def recommend(movie):
    movie_index = cleaned_movies[cleaned_movies['title'] == movie].index[0]
    distances = similarity[movie_index]
    movies_list = sorted(list(enumerate(distances)),reverse=True,key=lambda x:x[1])[1:6]

    for i in movies_list:
        print(cleaned_movies.iloc[i[0]].title)
```

```python
recommend('Avatar')
```
